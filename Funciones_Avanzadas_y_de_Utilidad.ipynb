**21. chr()**

**Uso **: Convierte un número entero en su carácter Unicode correspondiente.

```python
# Ejemplo de chr()
print("El carácter para el código 65 es:", chr(65))  # Resultado: 'A'
```

**22. ord()**

**Uso**: Devuelve el valor entero que representa el código Unicode de un carácter específico.

```python
# Ejemplo de ord()
print("El código Unicode de 'A' es:", ord('A'))  # Resultado: 65
```

**23. hex()**

Uso **texto en negrita**: Convierte un número entero en una cadena de texto con su representación en formato hexadecimal.

```python
# Ejemplo de hex()
print("El número 255 en hexadecimal es:", hex(255))  # Resultado: '0xff'
```

**24. bin()**

**Uso**: Convierte un número entero en su representación binaria en formato de cadena de texto.

```python
# Ejemplo de bin()
print("El número 5 en binario es:", bin(5))  # Resultado: '0b101'
```

**25. oct()**

**Uso**: Convierte un número entero en su representación octal en formato de cadena de texto.

```python
# Ejemplo de oct()
print("El número 8 en octal es:", oct(8))  # Resultado: '0o10'
```

**26. all()**

**Uso**: Devuelve True si todos los elementos de un iterable son verdaderos (o si está vacío).

```python
# Ejemplo de all()
valores = [True, True, 5 > 2]
print("¿Son todos verdaderos?:", all(valores))  # Resultado: True
```

**27. any()**

**Uso**: Devuelve True si al menos uno de los elementos de un iterable es verdadero.



```python
# Ejemplo de any()
valores = [False, False, True]
print("¿Hay al menos uno verdadero?:", any(valores))  # Resultado: True
```

**28. enumerate()**

**Uso**: Agrega un contador a un iterable y lo devuelve como un objeto enumerado, muy útil para obtener el índice y el valor en bucles.

```python
# Ejemplo de enumerate()
frutas = ["manzana", "pera", "uva"]
for indice, fruta in enumerate(frutas):
    print(indice, fruta)
```

**29. zip()**

**Uso**: Agrupa elementos de dos o más iterables (como listas) elemento por elemento en pares dentro de tuplas.

```python
# Ejemplo de zip()
nombres = ["Ana", "Carlos", "Sofía"]
edades = [19, 21, 20]
combinados = list(zip(nombres, edades))
print(combinados)
```

**30. reversed()**

**Uso**: Devuelve un iterador invertido de una secuencia dada.

```python
# Ejemplo de reversed()
numeros = [1, 2, 3, 4]
invertidos = list(reversed(numeros))
print("Lista invertida:", invertidos)  # Resultado: [4, 3, 2, 1]
```
