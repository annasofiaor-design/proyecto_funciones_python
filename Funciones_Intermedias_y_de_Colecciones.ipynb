**11. max()**

**Uso**: Devuelve el valor más alto de un iterable (como una lista o tupla) o el mayor de dos o más argumentos.

```python
# Ejemplo de max()
precios = [1200, 4500, 800, 3100]
print("El precio máximo es:", max(precios))
```

**12. min()**

**Uso**: Devuelve el valor más bajo de un iterable o el menor entre varios argumentos.

```python
# Ejemplo de min()
precios = [1200, 4500, 800, 3100]
print("El precio mínimo es:", min(precios))
```

**13. pow()**

**Uso**: Calcula la potencia de un número elevando la base al exponente indicado (base^exponente).

```python
# Ejemplo de pow() (2 elevado a la 3)
resultado = pow(2, 3)
print("2 elevado a la 3 es:", resultado)
```

14. bool()

**Uso**: Convierte un valor en un tipo de dato booleano (True o False) basándose en su evaluación lógica.

```python
# Ejemplo de bool()
print(bool(1))
print(bool(""))
```

**15. float()**

**Uso:** Convierte un número entero o una cadena de texto numérica en un número decimal (flotante).

```python
# Ejemplo de float()
numero_entero = 7
print(float(numero_entero))
```

**16. list()**

**Uso**: Crea o convierte un objeto iterable (como una tupla o un rango) en una lista modificable.

```python
# Ejemplo de list()
elementos = range(1, 20)
mi_lista = list(elementos)
print("Lista convertida:", mi_lista)
```

**17. set()**

**Uso**: Crea un conjunto a partir de un iterable. Los conjuntos eliminan automáticamente los elementos duplicados.

```python
# Ejemplo de set() con elementos repetidos
colores = ["rojo", "azul", "rojo", "verde", "azul"]
colores_unicos = set(colores)
print("Colores sin duplicados:", colores_unicos)
```

**18. dict()**

**Uso**: Se utiliza para crear un diccionario (una colección de pares clave-valor).

```python
# Ejemplo de dict()
estudiante = dict(nombre="Ana Sofía", edad=19, carrera="Ingeniería")
print(estudiante)
```

**19. sorted()**

**Uso**: Retorna una nueva lista ordenada con los elementos de cualquier iterable, ya sea de forma ascendente o descendente.   Ejemplo práctico (Código):

```python
# Ejemplo de sorted()
numeros_desordenados = [5, 2, 9, 1, 5, 6]
print("Lista ordenada:", sorted(numeros_desordenados))
```

**20. help()**

**Uso** : Invoca el sistema de ayuda interactivo de Python para consultar la documentación de funciones, módulos u objetos.

```python
# Ejemplo de help() (muestra información sobre la función sum)
help(sum)
```
