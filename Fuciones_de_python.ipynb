**1.funcion print**

**Uso**: Se utiliza para mostrar mensajes, variables o resultados en la pantalla o consola.

```python
# Ejemplo de print()
mensaje = "Hola, bienvenidos a la investigación de Python"
print(mensaje)
```

**2. Función len()**

**Uso:** Devuelve la longitud o el número total de elementos de un objeto (como una cadena de texto, lista o tupla).

```python
# Ejemplo de len()
texto = "Ciencia de Datos"
print("La longitud del texto es:", len(texto))
```

**3. Función type()**

**Uso**: Devuelve el tipo de dato del objeto o variable que se le pasa como parámetro (ej. int, str, float).

```python
# Ejemplo de type()
numero = 25.5
print("El tipo de dato es:", type(numero))
```

**4. Función input()**

**Uso:** Permite capturar información ingresada por el usuario a través del teclado en la consola.

```python
# Ejemplo de input()
nombre = input("Ingresa tu nombre: ")
print("Hola,",nombre )
```

**5. Función int()**

 **Uso**:Convierte un valor (como un número decimal o un texto numérico) a un número entero.

```python
# Ejemplo de int()
texto_numero = "100"
entero= int(texto_numero)
print(entero + 50)
```

**6. Función str()**

**Uso**: Convierte cualquier tipo de dato (números, booleanos, etc.) en una cadena de texto (string).

```python
# Ejemplo de str()
edad = 20
print("Mi edad es " + str(edad) + " años.")
```

**7. Función range()**

**Uso**: Genera una secuencia numérica estandarizada, muy utilizada para controlar bucles o iteraciones.

```python
# Ejemplo de range()
numeros = list(range(1, 20))
print("Lista de números:", numeros)
```

**8. Función abs()**

**Uso:** Retorna el valor absoluto de un número, eliminando cualquier signo negativo.

```python
# Ejemplo de abs()
numero_negativo = -15.8
print("El valor absoluto es:", abs(numero_negativo))
```

**9. Función round()**

**Uso**: Redondea un número decimal al entero más cercano o a una cantidad específica de decimales.

```python
# Ejemplo de round()
pi = 3.14159265
print("Pi redondeado a 2 decimales:", round(pi, 2))
```

**10. Función sum()**

**Uso**: Suma todos los elementos numéricos que forman parte de un contenedor iterable (como una lista).

```python
# Ejemplo de sum()
calificaciones = [4.5, 3.8, 5.0, 4.2]
print("La suma de las notas es:", sum(calificaciones))
```
